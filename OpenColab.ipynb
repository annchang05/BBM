```python
#@title Step 2: Import datasets & models from GitHub
import os
import shutil
import subprocess
from pathlib import Path

# Repositories to clone
repos = [
    "https://github.com/opensim-org/opensim-models.git",
    "https://github.com/HernandezVincent/OpenSim.git",
    "https://github.com/ESJiang/2354_xml.git",
    "https://github.com/ESJiang/2354_result.git",
    "https://github.com/ESJiang/testc3d.git"
]

# Clone repositories if they don't already exist locally
for repo_url in repos:
    folder_name = repo_url.split("/")[-1].replace(".git", "")
    if not os.path.exists(folder_name):
        print(f"Cloning {folder_name}...")
        try:
            subprocess.run(["git", "clone", repo_url], check=True)
        except subprocess.CalledProcessError:
            print(f" Warning: Could not clone {repo_url}. Skipping...")
    else:
        print(f"Directory '{folder_name}' already exists. Skipping clone.")

# Define local target paths
target_dir = Path("opensim-models/Pipelines/Gait2354_Simbody")
src_gait = Path("opensim-models/Models/Gait2354_Simbody")
src_xml = Path("2354_xml")

# Ensure target directory exists
target_dir.mkdir(parents=True, exist_ok=True)

# Copy gait files
if src_gait.exists():
    for file in src_gait.glob("*"):
        if file.is_file():
            shutil.copy(file, target_dir)

# Copy xml files
if src_xml.exists():
    for item in src_xml.glob("*"):
        if item.is_file():
            shutil.copy(item, target_dir)
        elif item.is_dir():
            dest_item = target_dir / item.name
            if dest_item.exists():
                shutil.rmtree(dest_item)
            shutil.copytree(item, dest_item)

print("Step 2 completed successfully!")
```

```python
# IK and ID visualization
# @title Step 1: OpenSim File Loader Function
from pathlib import Path
import pandas as pd

def read_opensim_file(file_path):
    """Parses an OpenSim .mot or .sto file into a pandas DataFrame
    by automatically detecting the header length.
    """
    file_path = Path(file_path)
    if not file_path.exists():
        raise FileNotFoundError(f"File not found: {file_path}")

    header_end_line = 0
    with open(file_path, "r") as f:
        for i, line in enumerate(f):
            if "endheader" in line.strip().lower():
                header_end_line = i + 1
                break

    # Read tab-delimited data following the header
    df = pd.read_csv(file_path, skiprows=header_end_line, sep=r"\s+")
    return df
```
